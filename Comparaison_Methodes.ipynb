```python
import numpy as np
import pandas as pd
```

```python
sup_users_suspects = pd.read_csv("supervise/predicted_users.csv")
sup_users_suspects = sup_users_suspects.loc[sup_users_suspects['atypique_pred'] == 1]
print("Nombre d'utilisateurs atypiques par SVM : ", sup_users_suspects.shape[0])
```

```python
non_sup_users_suspects = pd.read_csv("data/k_means_user.csv")
non_sup_users_suspects = non_sup_users_suspects.loc[non_sup_users_suspects['atypique_pred'] == 1]
print("Nombre d'utilisateurs atypiques par K_means : ", non_sup_users_suspects.shape[0])
```

```python
sup_ids = sup_users_suspects['user_id']
non_sup_ids = non_sup_users_suspects['user_id']
prop = np.sum(non_sup_ids.isin(sup_ids)) / len(non_sup_ids) * 100
print(f"{prop:.2f}% des utilisateurs atypiques selon k_means sont aussi atypiques selon le SVM")
```
